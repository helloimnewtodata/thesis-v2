```python
universe = [".STOXX",
            ".STOXXR"
]

fields = [
    "TR.PriceClose.date",
    "TR.PriceClose",
]

params = {
    "SDate": "2024-01-01",
    "EDate": "2026-03-05",
    "Frq": "D",
    "Curn": "EUR"
}

df_index = rd.get_data(
    universe=universe,
    fields=fields,
    parameters=params
)

df_index.head()
```

```python
# Varmista Date on datetime ilman kellonaikaa
df_index['Date'] = pd.to_datetime(df_index['Date']).dt.date

# Pudota mahdolliset duplikaatit
df_index = df_index.drop_duplicates(subset=['Date', 'Instrument'], keep='last')

df_pivot = df_index.pivot(index='Date', columns='Instrument', values='Price Close')
df_pivot
```

```python
# calculate the total return for the period for both indices
total_return_stoxx = (df_pivot['.STOXX'].iloc[-1] / df_pivot['.STOXX'].iloc[0]) - 1
total_return_stoxxr = (df_pivot['.STOXXR'].iloc[-1] / df_pivot['.STOXXR'].iloc[0]) - 1
```

```python
beta_df = pd.merge(
    df[['Date', 'Daily Total Return']].rename(columns={'Daily Total Return': 'NESTE'}),
    df_index.loc[df_index['Instrument'] == '.STOXXR', ['Date', 'Daily Return']].rename(columns={'Daily Return': 'STOXXR'}),
    on='Date',
    how='inner'
)

# Rolling beta (252 päivää)
cov = beta_df['NESTE'].rolling(252).cov(beta_df['STOXXR'])
var = beta_df['STOXXR'].rolling(252).var()
beta_df['Beta_252'] = cov / var

beta_df.set_index('Date', inplace=True)
beta_df['Beta_252'].plot(title='NESTE.HE Rolling 252d Beta vs .STOXXR', figsize=(12, 5))
```
